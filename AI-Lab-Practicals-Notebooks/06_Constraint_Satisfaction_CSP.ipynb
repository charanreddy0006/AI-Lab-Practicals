## Objective

## 1. Map coloring
Assign a color to every region so neighboring regions have different colors.

```python
variables = ["A", "B", "C", "D"]
domains = {v: ["Red", "Green", "Blue"] for v in variables}
neighbors = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

def is_consistent(variable, color, assignment):
    return all(assignment.get(n) != color for n in neighbors[variable])

def backtracking_map(assignment):
    if len(assignment) == len(variables):
        return assignment.copy()
    unassigned = [v for v in variables if v not in assignment]
    variable = unassigned[0]
    for color in domains[variable]:
        if is_consistent(variable, color, assignment):
            assignment[variable] = color
            result = backtracking_map(assignment)
            if result is not None:
                return result
            del assignment[variable]
    return None

solution = backtracking_map({})
print("Map coloring solution:", solution)
```

## 2. N-Queens
Place N queens on an N×N chessboard so no two queens attack each other.

```python
def solve_n_queens(n):
    placement = []  # placement[row] = column

    def safe(row, col):
        for prev_row, prev_col in enumerate(placement):
            if prev_col == col:
                return False
            if abs(prev_row - row) == abs(prev_col - col):
                return False
        return True

    def backtrack(row):
        if row == n:
            return placement.copy()
        for col in range(n):
            if safe(row, col):
                placement.append(col)
                result = backtrack(row + 1)
                if result is not None:
                    return result
                placement.pop()
        return None

    return backtrack(0)

n = 8
queens = solve_n_queens(n)
print(f"{n}-Queens solution (column for each row):", queens)
```

```python
def print_chessboard(queens):
    n = len(queens)
    for row in range(n):
        print(" ".join("Q" if queens[row] == col else "." for col in range(n)))

print_chessboard(queens)
```
