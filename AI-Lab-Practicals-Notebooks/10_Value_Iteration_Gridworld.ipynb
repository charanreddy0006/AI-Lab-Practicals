# Practical 10: Sequential Decision Making using Value Iteration

## Objective
Implement value iteration in a grid-world Markov Decision Process and extract a policy.

**Course:** Artificial Intelligence (01AI0511)  
**Language:** Python 3

> Run the notebook from top to bottom. The examples use small, readable datasets so the algorithms can be understood and modified.

```python
# Grid: S = start, G = goal, # = blocked cell
grid = [
    ["S", ".", "."],
    [".", "#", "."],
    [".", ".", "G"]
]
rows, cols = len(grid), len(grid[0])
states = [(r, c) for r in range(rows) for c in range(cols) if grid[r][c] != "#"]
goal = (2, 2)
actions = {"U": (-1,0), "D": (1,0), "L": (0,-1), "R": (0,1)}
gamma = 0.9
step_reward = -1.0
goal_reward = 10.0

def transition(state, action):
    dr, dc = actions[action]
    nxt = (state[0] + dr, state[1] + dc)
    if not (0 <= nxt[0] < rows and 0 <= nxt[1] < cols) or grid[nxt[0]][nxt[1]] == "#":
        nxt = state
    return nxt

def reward(state, next_state):
    return goal_reward if next_state == goal else step_reward
```

## 1. Value iteration
For this deterministic MDP, each action has one next state.

```python
def value_iteration(theta=1e-6, max_iterations=1000):
    values = {s: 0.0 for s in states}
    values[goal] = 0.0

    for iteration in range(max_iterations):
        delta = 0.0
        updated = values.copy()
        for state in states:
            if state == goal:
                continue
            candidates = []
            for action in actions:
                nxt = transition(state, action)
                candidates.append(reward(state, nxt) + gamma * values[nxt])
            updated[state] = max(candidates)
            delta = max(delta, abs(updated[state] - values[state]))
        values = updated
        if delta < theta:
            return values, iteration + 1
    return values, max_iterations

values, iterations = value_iteration()
print("Converged after", iterations, "iterations")
for r in range(rows):
    print([round(values[(r,c)], 2) if (r,c) in values else None for c in range(cols)])
```

## 2. Extract a greedy policy

```python
def extract_policy(values):
    policy = {}
    for state in states:
        if state == goal:
            policy[state] = "G"
            continue
        best_action = max(
            actions,
            key=lambda a: reward(state, transition(state, a))
                        + gamma * values[transition(state, a)]
        )
        policy[state] = best_action
    return policy

policy = extract_policy(values)
for r in range(rows):
    print(" ".join(policy.get((r,c), "#") for c in range(cols)))
```
