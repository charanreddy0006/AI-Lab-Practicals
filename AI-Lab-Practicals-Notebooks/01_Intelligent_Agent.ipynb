## 1. Simple environment
A small two-room world. Each room can be clean or dirty. The agent can move left/right or clean the current room.

```python
class VacuumWorld:
    def __init__(self, rooms=None, start="A"):
        self.rooms = rooms or {"A": "Dirty", "B": "Dirty"}
        self.position = start

    def percept(self):
        return self.position, self.rooms[self.position]

    def execute(self, action):
        if action == "CLEAN":
            self.rooms[self.position] = "Clean"
        elif action == "LEFT":
            self.position = "A"
        elif action == "RIGHT":
            self.position = "B"
        elif action != "NOOP":
            raise ValueError(f"Unknown action: {action}")
        return self.percept()

    def is_goal(self):
        return all(status == "Clean" for status in self.rooms.values())


world = VacuumWorld()
print("Initial state:", world.rooms, "| position:", world.position)
```

## 2. Simple reflex agent
The reflex agent chooses an action using only the current percept.

```python
def reflex_agent(percept):
    position, status = percept
    if status == "Dirty":
        return "CLEAN"
    return "RIGHT" if position == "A" else "LEFT"


world = VacuumWorld()
for step in range(6):
    percept = world.percept()
    action = reflex_agent(percept)
    print(f"Step {step + 1}: percept={percept}, action={action}")
    world.execute(action)
    if world.is_goal():
        print("Both rooms are clean.")
        break
print("Final rooms:", world.rooms)
```

## 3. Goal-based agent
This agent checks the overall goal and chooses actions to clean both rooms.

```python
def goal_based_agent(world):
    if world.is_goal():
        return "NOOP"
    if world.rooms[world.position] == "Dirty":
        return "CLEAN"
    if world.position == "A" and world.rooms["B"] == "Dirty":
        return "RIGHT"
    if world.position == "B" and world.rooms["A"] == "Dirty":
        return "LEFT"
    return "NOOP"


world = VacuumWorld({"A": "Clean", "B": "Dirty"}, start="A")
for step in range(5):
    action = goal_based_agent(world)
    print(f"Step {step + 1}: position={world.position}, rooms={world.rooms}, action={action}")
    world.execute(action)
    if world.is_goal():
        print("Goal reached.")
        break
```
