## 1. Propositional logic
Evaluate logical expressions using Boolean variables.

```python
# Example propositions
P = True   # It is raining
Q = False  # The ground is wet (assumed false for this example)

print("P AND Q:", P and Q)
print("P OR Q:", P or Q)
print("NOT P:", not P)
print("P implies Q:", (not P) or Q)
```

## 2. Truth table for implication

```python
def implies(p, q):
    return (not p) or q

print("P     Q     P -> Q")
for p in [False, True]:
    for q in [False, True]:
        print(f"{str(p):5} {str(q):5} {implies(p, q)}")
```

## 3. Simple first-order facts and rules
Facts are represented as tuples. The rule below expresses: if X is a parent of Y and Y is a parent of Z, then X is a grandparent of Z.

```python
facts = {
    ("parent", "alice", "bob"),
    ("parent", "bob", "carol"),
    ("parent", "alice", "david"),
    ("parent", "david", "emma")
}

def infer_grandparents(known_facts):
    parents = [(x, y) for relation, x, y in known_facts if relation == "parent"]
    inferred = set()
    for x, y in parents:
        for y2, z in parents:
            if y == y2:
                inferred.add(("grandparent", x, z))
    return inferred

inferred_facts = infer_grandparents(facts)
print("Inferred facts:")
for fact in sorted(inferred_facts):
    print(fact)
```

## 4. Query the knowledge base

```python
def query_relation(knowledge_base, relation, first, second):
    return (relation, first, second) in knowledge_base

knowledge_base = facts | inferred_facts
print("Is Alice a grandparent of Carol?",
      query_relation(knowledge_base, "grandparent", "alice", "carol"))
print("Is Bob a grandparent of Emma?",
      query_relation(knowledge_base, "grandparent", "bob", "emma"))
```

## Observation
Propositional logic works with truth-valued statements. First-order logic represents objects and relations, enabling more structured knowledge representation.
