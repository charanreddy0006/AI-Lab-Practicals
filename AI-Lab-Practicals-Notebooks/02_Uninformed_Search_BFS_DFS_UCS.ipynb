# Practical 2: Uninformed Search: BFS, DFS and Uniform-Cost Search

## Objective
Implement breadth-first search, depth-first search and uniform-cost search on a graph, then compare the paths and costs.

**Language:** Python 3

> Run the notebook from top to bottom. The examples use small, readable datasets so the algorithms can be understood and modified.

```python
from collections import deque
import heapq

graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("D", 2), ("E", 5)],
    "C": [("F", 1)],
    "D": [("G", 3)],
    "E": [("G", 1)],
    "F": [("G", 2)],
    "G": []
}
```

## 1. Breadth-first search
BFS explores nodes level by level and finds a path with the fewest edges in an unweighted graph.

```python
def bfs(graph, start, goal):
    queue = deque([(start, [start])])
    visited = {start}
    while queue:
        node, path = queue.popleft()
        if node == goal:
            return path
        for neighbor, _ in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))
    return None

bfs_path = bfs(graph, "A", "G")
print("BFS path:", bfs_path)
```

## 2. Depth-first search
DFS explores one branch as far as possible before backtracking.

```python
def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()
    while stack:
        node, path = stack.pop()
        if node == goal:
            return path
        if node in visited:
            continue
        visited.add(node)
        for neighbor, _ in reversed(graph[node]):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))
    return None

dfs_path = dfs(graph, "A", "G")
print("DFS path:", dfs_path)
```

## 3. Uniform-cost search
UCS expands the frontier node with the lowest accumulated path cost.

```python
def uniform_cost_search(graph, start, goal):
    frontier = [(0, start, [start])]
    best_cost = {start: 0}
    while frontier:
        cost, node, path = heapq.heappop(frontier)
        if cost != best_cost.get(node):
            continue
        if node == goal:
            return path, cost
        for neighbor, edge_cost in graph[node]:
            new_cost = cost + edge_cost
            if new_cost < best_cost.get(neighbor, float("inf")):
                best_cost[neighbor] = new_cost
                heapq.heappush(frontier, (new_cost, neighbor, path + [neighbor]))
    return None, float("inf")

ucs_path, ucs_cost = uniform_cost_search(graph, "A", "G")
print("UCS path:", ucs_path)
print("UCS total cost:", ucs_cost)
```

```python
def path_cost(graph, path):
    if not path:
        return None
    total = 0
    for a, b in zip(path, path[1:]):
        total += next(cost for neighbor, cost in graph[a] if neighbor == b)
    return total

print("BFS path cost:", path_cost(graph, bfs_path))
print("DFS path cost:", path_cost(graph, dfs_path))
print("UCS path cost:", ucs_cost)
```

## Observation
BFS minimizes the number of edges, while UCS minimizes total edge cost. DFS depends strongly on neighbor ordering.
