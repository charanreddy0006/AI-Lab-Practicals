# Practical 3: A* Heuristic Search for Grid Path Planning

## Objective
Implement A* search for a grid path-planning problem and evaluate how a heuristic guides the search.

**Course:** Artificial Intelligence (01AI0511)  
**Language:** Python 3

> Run the notebook from top to bottom. The examples use small, readable datasets so the algorithms can be understood and modified.

```python
import heapq

grid = [
    "S....",
    ".##..",
    "...#.",
    ".#...",
    "....G"
]
rows, cols = len(grid), len(grid[0])

def find_cell(symbol):
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == symbol:
                return (r, c)
    raise ValueError(f"{symbol} not found")

start, goal = find_cell("S"), find_cell("G")

def manhattan(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def neighbors(pos):
    r, c = pos
    for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] != "#":
            yield (nr, nc)

def reconstruct_path(came_from, current):
    path = [current]
    while current in came_from:
        current = came_from[current]
        path.append(current)
    return list(reversed(path))
```

```python
def astar(start, goal, heuristic=manhattan):
    frontier = [(heuristic(start, goal), 0, start)]
    came_from = {}
    g_score = {start: 0}
    expanded = 0

    while frontier:
        _, cost, current = heapq.heappop(frontier)
        if cost != g_score.get(current):
            continue
        expanded += 1
        if current == goal:
            return reconstruct_path(came_from, current), cost, expanded

        for nxt in neighbors(current):
            tentative = cost + 1
            if tentative < g_score.get(nxt, float("inf")):
                came_from[nxt] = current
                g_score[nxt] = tentative
                priority = tentative + heuristic(nxt, goal)
                heapq.heappush(frontier, (priority, tentative, nxt))
    return None, float("inf"), expanded

path, cost, expanded = astar(start, goal)
print("Path:", path)
print("Path length:", cost)
print("Expanded nodes:", expanded)
```

```python
def display_grid(path):
    path_set = set(path or [])
    for r in range(rows):
        line = ""
        for c in range(cols):
            p = (r, c)
            if p == start:
                line += "S "
            elif p == goal:
                line += "G "
            elif grid[r][c] == "#":
                line += "# "
            elif p in path_set:
                line += "* "
            else:
                line += ". "
        print(line)

display_grid(path)
```
