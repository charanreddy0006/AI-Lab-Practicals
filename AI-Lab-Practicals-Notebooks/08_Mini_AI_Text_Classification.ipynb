## 1. Prepare a tiny labelled dataset
This self-contained example uses keyword features and a transparent scoring rule.

```python
training_data = [
    ("win a free prize now", "spam"),
    ("claim your reward today", "spam"),
    ("limited offer click here", "spam"),
    ("meeting moved to tomorrow", "ham"),
    ("please send the project notes", "ham"),
    ("are we still meeting today", "ham"),
    ("free cash prize claim now", "spam"),
    ("the class starts at nine", "ham")
]

spam_words = {"free", "prize", "claim", "offer", "cash", "win", "click", "reward"}
ham_words = {"meeting", "project", "class", "notes", "tomorrow", "today", "starts"}
```

## 2. Train a simple keyword model

```python
def tokenize(text):
    return [word.strip(".,!?").lower() for word in text.split()]

def predict_message(text):
    words = tokenize(text)
    spam_score = sum(word in spam_words for word in words)
    ham_score = sum(word in ham_words for word in words)
    label = "spam" if spam_score > ham_score else "ham"
    if spam_score == ham_score:
        label = "uncertain"
    return label, {"spam_score": spam_score, "ham_score": ham_score}

tests = [
    "claim your free prize",
    "please send the meeting notes",
    "the project meeting is tomorrow"
]
for message in tests:
    print(message, "=>", predict_message(message))
```

## 3. Evaluate on the small training set
This is a demonstration dataset, not a reliable production benchmark.

```python
correct = 0
for text, expected in training_data:
    predicted, _ = predict_message(text)
    correct += predicted == expected
accuracy = correct / len(training_data)
print(f"Training-set accuracy: {accuracy:.1%} ({correct}/{len(training_data)})")
```

## 4. Explainability and responsible AI

```python
sample = "free prize meeting tomorrow"
label, scores = predict_message(sample)
print("Text:", sample)
print("Prediction:", label)
print("Feature scores:", scores)
print("\\nLimitations:")
print("- Very small dataset; results may not generalize.")
print("- Keyword rules can be biased by vocabulary and context.")
print("- Do not use this demo for high-impact decisions.")
print("- Review errors and protect private text data.")
```
