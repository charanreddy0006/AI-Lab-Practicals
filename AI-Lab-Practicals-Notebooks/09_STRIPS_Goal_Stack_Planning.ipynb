# Practical 9: Planning using STRIPS-Style Actions

## Objective
Represent a small Blocks World planning problem using preconditions, add effects and delete effects, then search for a plan.

## 1. Represent the planning problem
A state is a set of facts. An action is applicable when all its preconditions are true.

```python
from collections import deque

initial_state = frozenset({
    ("ontable", "A"), ("ontable", "B"),
    ("clear", "A"), ("clear", "B"),
    ("handempty",)
})
goal_facts = frozenset({("on", "A", "B"), ("ontable", "B")})

actions = [
    {
        "name": "PICKUP_A",
        "pre": {("ontable", "A"), ("clear", "A"), ("handempty",)},
        "add": {("holding", "A")},
        "delete": {("ontable", "A"), ("clear", "A"), ("handempty",)}
    },
    {
        "name": "STACK_A_ON_B",
        "pre": {("holding", "A"), ("clear", "B")},
        "add": {("on", "A", "B"), ("clear", "A"), ("handempty",)},
        "delete": {("holding", "A"), ("clear", "B")}
    }
]
```

## 2. Apply an action

```python
def apply_action(state, action):
    if not action["pre"].issubset(state):
        return None
    new_state = set(state)
    new_state.difference_update(action["delete"])
    new_state.update(action["add"])
    return frozenset(new_state)

for action in actions:
    result = apply_action(initial_state, action)
    print(action["name"], "applicable:", result is not None)
```

## 3. Search for a plan with breadth-first search

```python
def goal_test(state):
    return goal_facts.issubset(state)

def bfs_plan(initial):
    queue = deque([(initial, [])])
    visited = {initial}
    while queue:
        state, plan = queue.popleft()
        if goal_test(state):
            return plan, state
        for action in actions:
            next_state = apply_action(state, action)
            if next_state is not None and next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, plan + [action["name"]]))
    return None, None

plan, final_state = bfs_plan(initial_state)
print("Plan:", plan)
print("Goal reached:", final_state is not None and goal_test(final_state))
print("Final state:", final_state)
```

## Observation
The planner finds a sequence of applicable actions that transforms the initial state into one satisfying the goal facts. This is a compact STRIPS-style example.
