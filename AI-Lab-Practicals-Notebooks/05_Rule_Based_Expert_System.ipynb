# Practical 5: Rule-Based Expert System

## Objective
Build a small rule-based expert system that uses facts and IF-THEN rules to provide study-support recommendations.

**Course:** Artificial Intelligence (01AI0511)  
**Language:** Python 3

> Run the notebook from top to bottom. The examples use small, readable datasets so the algorithms can be understood and modified.

```python
# Facts supplied by a user (edit these values to test other cases).
facts = {
    "has_fever": True,
    "has_cough": True,
    "has_sore_throat": False,
    "has_low_energy": True,
    "needs_rest": False
}

# Rules are deliberately non-medical and demonstrate inference only.
rules = [
    ({"has_fever": True, "has_cough": True},
     "Consider taking a break and seeking advice from a qualified professional if unwell."),
    ({"has_low_energy": True},
     "Consider scheduling a rest break and reviewing your workload."),
    ({"has_sore_throat": True},
     "Keep track of symptoms and seek appropriate professional advice if needed.")
]
```

## 1. Forward chaining
A rule fires when all its conditions match the known facts.

```python
def forward_chain(facts, rules):
    conclusions = []
    for conditions, conclusion in rules:
        if all(facts.get(key) == expected for key, expected in conditions.items()):
            conclusions.append(conclusion)
    return conclusions

results = forward_chain(facts, rules)
print("Known facts:")
for key, value in facts.items():
    print(f"  {key}: {value}")
print("\\nExpert-system recommendations:")
for item in results:
    print("-", item)
```

## 2. Explain which rules fired

```python
def explain_rules(facts, rules):
    explanations = []
    for index, (conditions, conclusion) in enumerate(rules, start=1):
        matched = all(facts.get(k) == v for k, v in conditions.items())
        explanations.append({
            "rule": index,
            "conditions": conditions,
            "fired": matched,
            "conclusion": conclusion if matched else None
        })
    return explanations

for explanation in explain_rules(facts, rules):
    print(explanation)
```

## 3. Add a custom rule
Add a new rule and run the inference engine again.

```python
rules.append((
    {"has_low_energy": True, "needs_rest": False},
    "Plan a short break and reassess how you feel."
))
for recommendation in forward_chain(facts, rules):
    print("-", recommendation)
```
