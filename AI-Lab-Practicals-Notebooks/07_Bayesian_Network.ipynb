## 1. Define the network
This example uses the classic Burglary–Earthquake–Alarm network. The conditional probability tables are illustrative.

```python
# P(Burglary), P(Earthquake)
P_B = 0.001
P_E = 0.002

# P(Alarm=True | Burglary, Earthquake)
P_A = {
    (True, True): 0.95,
    (True, False): 0.94,
    (False, True): 0.29,
    (False, False): 0.001
}

# P(JohnCalls=True | Alarm)
P_J_given_A = {True: 0.90, False: 0.05}
# P(MaryCalls=True | Alarm)
P_M_given_A = {True: 0.70, False: 0.01}
```

## 2. Joint probability

```python
def bernoulli_probability(value, p_true):
    return p_true if value else 1 - p_true

def joint_probability(b, e, a, j, m):
    p = bernoulli_probability(b, P_B)
    p *= bernoulli_probability(e, P_E)
    p *= bernoulli_probability(a, P_A[(b, e)])
    p *= bernoulli_probability(j, P_J_given_A[a])
    p *= bernoulli_probability(m, P_M_given_A[a])
    return p

print("Example joint probability:",
      joint_probability(True, False, True, True, False))
```

## 3. Query P(Burglary | JohnCalls=True, MaryCalls=True)
Enumerate all hidden variables and normalize the two cases for Burglary.

```python
def probability_burglary_given_calls():
    weights = {True: 0.0, False: 0.0}
    for b in [True, False]:
        for e in [True, False]:
            for a in [True, False]:
                # Evidence: both John and Mary call.
                weights[b] += joint_probability(b, e, a, True, True)
    total = sum(weights.values())
    return {b: weight / total for b, weight in weights.items()}

posterior = probability_burglary_given_calls()
print("P(Burglary=True | both call) =", round(posterior[True], 6))
print("P(Burglary=False | both call) =", round(posterior[False], 6))
```

## 4. Interpretation
The posterior probability is higher than the prior because calls from both neighbors are evidence that an alarm may have occurred. The result depends on the chosen probability tables.
