# Practical 4: Game-Playing Agent using Minimax and Alpha-Beta Pruning

## Objective
Develop a Tic-Tac-Toe game-playing agent using minimax and alpha-beta pruning.

**Course:** Artificial Intelligence (01AI0511)  
**Language:** Python 3

> Run the notebook from top to bottom. The examples use small, readable datasets so the algorithms can be understood and modified.

```python
EMPTY = " "
WIN_LINES = [
    (0,1,2), (3,4,5), (6,7,8),
    (0,3,6), (1,4,7), (2,5,8),
    (0,4,8), (2,4,6)
]

def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] != EMPTY and board[a] == board[b] == board[c]:
            return board[a]
    return None

def terminal_score(board):
    result = winner(board)
    if result == "X": return 1
    if result == "O": return -1
    if EMPTY not in board: return 0
    return None

def available_moves(board):
    return [i for i, value in enumerate(board) if value == EMPTY]

def show_board(board):
    for i in range(0, 9, 3):
        print(" | ".join(board[i:i+3]))
        if i < 6: print("--+---+--")
```

## 1. Minimax algorithm
X is the maximizing player and O is the minimizing player.

```python
def minimax(board, maximizing):
    score = terminal_score(board)
    if score is not None:
        return score, None

    if maximizing:
        best_score, best_move = -float("inf"), None
        for move in available_moves(board):
            board[move] = "X"
            score, _ = minimax(board, False)
            board[move] = EMPTY
            if score > best_score:
                best_score, best_move = score, move
        return best_score, best_move
    else:
        best_score, best_move = float("inf"), None
        for move in available_moves(board):
            board[move] = "O"
            score, _ = minimax(board, True)
            board[move] = EMPTY
            if score < best_score:
                best_score, best_move = score, move
        return best_score, best_move

board = [EMPTY] * 9
score, move = minimax(board, True)
print("Best opening move for X:", move, "| score:", score)
```

## 2. Minimax with alpha-beta pruning
Alpha and beta bound the values that can still affect the final decision, allowing some branches to be skipped.

```python
def minimax_alpha_beta(board, maximizing, alpha=-float("inf"), beta=float("inf"), stats=None):
    if stats is None:
        stats = {"visited": 0, "pruned": 0}
    stats["visited"] += 1
    score = terminal_score(board)
    if score is not None:
        return score, None, stats

    best_move = None
    if maximizing:
        best_score = -float("inf")
        for move in available_moves(board):
            board[move] = "X"
            score, _, stats = minimax_alpha_beta(board, False, alpha, beta, stats)
            board[move] = EMPTY
            if score > best_score:
                best_score, best_move = score, move
            alpha = max(alpha, best_score)
            if beta <= alpha:
                stats["pruned"] += 1
                break
        return best_score, best_move, stats
    else:
        best_score = float("inf")
        for move in available_moves(board):
            board[move] = "O"
            score, _, stats = minimax_alpha_beta(board, True, alpha, beta, stats)
            board[move] = EMPTY
            if score < best_score:
                best_score, best_move = score, move
            beta = min(beta, best_score)
            if beta <= alpha:
                stats["pruned"] += 1
                break
        return best_score, best_move, stats

board = [EMPTY] * 9
score, move, stats = minimax_alpha_beta(board, True)
print("Best move:", move, "| score:", score)
print("Search statistics:", stats)
```

## 3. Play one move
The following shows the AI's move from an empty board.

```python
board = [EMPTY] * 9
_, ai_move, _ = minimax_alpha_beta(board, True)
board[ai_move] = "X"
show_board(board)
```
